# LLM Routing vs Semantic Routing

LLM Routing uses LLM inference to decide where to route requests, which is slow and expensive. Semantic Routing uses embedding similarities, offering much faster, cheaper routing.

```
import numpy as np

def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

print(cosine_similarity([1, 0], [0.8, 0.6]))
```
