# The Routing Problem: Latency, Cost, and Accuracy

Large Language Models (LLMs) present a trilemma: cost, latency, and accuracy. Fast models are cheap but inaccurate. Accurate models are slow and expensive. 

To optimize this, we need a routing mechanism to send easy queries to fast models and hard queries to complex models. This routing relies on embeddings and similarity search, typically Cosine Similarity.


```
# Raw Cosine Similarity from scratch
import math

def dot_product(v1, v2):
    return sum(x * y for x, y in zip(v1, v2))

def magnitude(v):
    return math.sqrt(sum(x * x for x in v))

def cosine_similarity(v1, v2):
    return dot_product(v1, v2) / (magnitude(v1) * magnitude(v2))

v_query = [0.1, 0.3, 0.5]
v_route_a = [0.1, 0.4, 0.4]
v_route_b = [0.8, 0.1, 0.1]

print("Sim A:", cosine_similarity(v_query, v_route_a))
print("Sim B:", cosine_similarity(v_query, v_route_b))
```
