# Architecture and O(N) Lazy Compilation

SynaptoRoute uses an O(N) lazy compilation tradeoff. It defers compiling embeddings until query time or a background process does it, keeping hot-reloads extremely fast.

```
from synaptoroute import AdaptiveRouter

router = AdaptiveRouter()
router.add_route('greeting', ['hello', 'hi'])
# Hot-reload an utterance
router.add_utterance('greeting', 'hey there')
print('Route and utterance added!')
```
